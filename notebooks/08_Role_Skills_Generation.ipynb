```python
import pandas as pd
import json
import re
from collections import defaultdict
```

```python
df = pd.read_csv(r"C:\Users\geeth\OneDrive\Desktop\ResumeIQ-AI-Resume-Screening-System\data\raw\ml_resume_dataset_4500.csv")

df.head()
```

```python
print(df.columns)

print(df.shape)

df.head()
```

```python
skills_df = df[["current_title", "skills"]].copy()

skills_df.head()
```

```python
df[["current_title", "label"]].head(20)
```

```python
print(df["current_title"].nunique())
print(df["label"].nunique())
```

```python
skills_df["skills"] = (
    skills_df["skills"]
    .str.lower()
    .str.replace(";", ",", regex=False)
    .str.replace("|", ",", regex=False)
)

skills_df.head()
```

```python
role_skills = defaultdict(set)

for _, row in skills_df.iterrows():

    role = row["current_title"]

    skills = row["skills"].split(",")

    for skill in skills:

        skill = skill.strip()

        if skill:

            role_skills[role].add(skill)
```

```python
role_skills = {

    role: sorted(list(skills))

    for role, skills in role_skills.items()

}

print("Total Roles:", len(role_skills))
```

```python
for role in list(role_skills.keys())[:10]:

    print("\n", role)

    print(role_skills[role][:15])
```

```python
output_path = "../data/processed/role_skills.json"

with open(output_path, "w") as file:

    json.dump(role_skills, file, indent=4)

print("Role skills saved successfully!")
```
