### Objective

Optimize the best-performing machine learning model using GridSearchCV.

### Goals

- Hyperparameter Tuning

```python
import joblib
import pandas as pd

from sklearn.svm import LinearSVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

print("Libraries Imported Successfully!")
```

```python
X_train = joblib.load(r"C:\Users\geeth\OneDrive\Desktop\ResumeIQ-AI-Resume-Screening-System\trained_models\X_train.pkl")
X_test = joblib.load(r"C:\Users\geeth\OneDrive\Desktop\ResumeIQ-AI-Resume-Screening-System\trained_models\X_test.pkl")

y_train = joblib.load(r"C:\Users\geeth\OneDrive\Desktop\ResumeIQ-AI-Resume-Screening-System\trained_models\y_train.pkl")
y_test = joblib.load(r"C:\Users\geeth\OneDrive\Desktop\ResumeIQ-AI-Resume-Screening-System\trained_models\y_test.pkl")
```

```python
baseline_model = LinearSVC(random_state=42)

baseline_model.fit(X_train, y_train)

baseline_predictions = baseline_model.predict(X_test)

baseline_accuracy = accuracy_score(y_test, baseline_predictions)

print(f"Baseline Accuracy : {baseline_accuracy:.4f}")
```

```python
parameter_grid = {
    "C": [0.01, 0.1, 1, 10, 100]
}
```

```python
grid_search = GridSearchCV(
    estimator=LinearSVC(random_state=42),
    param_grid=parameter_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross Validation Accuracy:")
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

predictions = best_model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)

print(f"Optimized Model Accuracy : {accuracy:.4f}")
```

```python
comparison = pd.DataFrame({
    "Model": ["Baseline LinearSVC", "Optimized LinearSVC"],
    "Accuracy": [baseline_accuracy, accuracy]
})

comparison
```

```python
joblib.dump(
    best_model,
    "../trained_models/optimized_resume_classifier.pkl"
)

print("Optimized Model Saved Successfully!")
```
