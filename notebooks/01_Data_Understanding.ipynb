```python
plt.figure(figsize=(12, 8))

sns.countplot(
    data=resume_df,
    y="Category",
    order=resume_df["Category"].value_counts().index
)

plt.title("Resume Category Distribution")
plt.xlabel("Count")
plt.ylabel("Category")

plt.show()
```

```python
duplicate_count = resume_df.duplicated().sum()

print(f"Duplicate Records : {duplicate_count}")
```

```python
resume_df["Resume_Length"] = resume_df["Resume"].apply(len)

resume_df["Resume_Length"].describe()
```

```python
plt.figure(figsize=(10,5))

plt.hist(resume_df["Resume_Length"], bins=30)

plt.title("Distribution of Resume Length")
plt.xlabel("Resume Length")
plt.ylabel("Frequency")

plt.show()
```

```python
from wordcloud import WordCloud

text = " ".join(resume_df["Resume"])

wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color="white"
).generate(text)

plt.figure(figsize=(15,8))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
