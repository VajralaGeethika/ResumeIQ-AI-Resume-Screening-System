# ResumeIQ - AI Resume Screening System

## Notebook 03: Feature Engineering

### Objective
Convert cleaned resume text into numerical features for machine learning.

### Steps
- Load cleaned dataset
- Encode target labels
- TF-IDF Vectorization
- Train-Test Split

```python
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

import joblib

print("Libraries imported successfully!")
```

```python

resume_df = pd.read_csv(r"C:\Users\geeth\OneDrive\Desktop\ResumeIQ-AI-Resume-Screening-System\data\processed\cleaned_resume_dataset.csv")

resume_df.head()
```

```python
resume_df.info()
```

```python
X = resume_df["Clean_Resume"]

y = resume_df["Category"]

print(X.shape)
print(y.shape)
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Number of Categories:", len(label_encoder.classes_))
print(label_encoder.classes_)
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2),
    stop_words="english"
)

X_tfidf = tfidf.fit_transform(X)

print("Feature Matrix Shape:", X_tfidf.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

print("Training Samples :", X_train.shape)
print("Testing Samples  :", X_test.shape)
```

```python
import joblib
joblib.dump(tfidf, "../trained_models/tfidf_vectorizer.pkl")
joblib.dump(label_encoder, "../trained_models/label_encoder.pkl")
joblib.dump(X_train, "../trained_models/X_train.pkl")
joblib.dump(X_test, "../trained_models/X_test.pkl")

joblib.dump(y_train, "../trained_models/y_train.pkl")
joblib.dump(y_test, "../trained_models/y_test.pkl")

print("Training and testing datasets saved successfully!")

print("Vectorizer and Label Encoder saved successfully!")
```
